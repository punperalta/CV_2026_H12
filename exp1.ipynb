```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("flower.jpg")

if image is None:
    print("Image not found!")
else:
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.imshow(image_rgb)
    plt.axis("off")
    plt.show()
```

```python
height, width, channels = image.shape
print((height, width, channels))
print(image.dtype)
```

```python
start_row, end_row = 300, 980
start_col, end_col = 100, 759

cropped_img = image[start_row:end_row, start_col:end_col]

plt.imshow(cv2.cvtColor(cropped_img, cv2.COLOR_BGR2RGB))
plt.title("Cropped Image")
plt.axis("off")
plt.show()
```

```python
new_width = 300
new_height = 500

scaled_img = cv2.resize(image, (new_width, new_height))

plt.imshow(cv2.cvtColor(scaled_img, cv2.COLOR_BGR2RGB))
plt.title("Resized Image")
plt.axis("off")
plt.show()
```

```python
edited_img = cv2.imread("flower.jpg")

row = 120
column = 100

edited_img[row, column] = (0, 0, 0)

plt.imshow(cv2.cvtColor(edited_img, cv2.COLOR_BGR2RGB))
plt.title("Modified Pixel")
plt.axis("off")
plt.show()
```

```python
gray_image = cv2.cvtColor(edited_img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray_image, cmap="gray")
plt.title("Gray Image")
plt.axis("off")
plt.show()
```

```python
sample = cv2.imread("flower.jpg")

rgb_version = cv2.cvtColor(sample, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10,4))

plt.subplot(121)
plt.imshow(sample)
plt.title("BGR")

plt.subplot(122)
plt.imshow(rgb_version)
plt.title("RGB")

plt.tight_layout()
plt.show()
```

```python
red = sample[:, :, 2]

plt.imshow(red, cmap="gray")
plt.title("Red Channel")
plt.axis("off")
plt.show()
```

```python
mirror = cv2.flip(sample, 1)

plt.imshow(cv2.cvtColor(mirror, cv2.COLOR_BGR2RGB))
plt.title("Mirror Image")
plt.axis("off")
plt.show()
```

```python
clockwise = cv2.rotate(sample, cv2.ROTATE_90_CLOCKWISE)

plt.imshow(cv2.cvtColor(clockwise, cv2.COLOR_BGR2RGB))
plt.title("Rotated Image")
plt.axis("off")
plt.show()
```

```python
smooth = cv2.GaussianBlur(sample, (15, 15), sigmaX=0)

plt.imshow(cv2.cvtColor(smooth, cv2.COLOR_BGR2RGB))
plt.title("Blurred Image")
plt.axis("off")
plt.show()
```
